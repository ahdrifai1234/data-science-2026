```python
# Membuat list
data = [10, 20, 'Python', True, 3.14]
# Akses & slicing
print(data[0]) # → 10
print(data[-1]) # → 3.14 (dari belakang)
print(data[1:3]) # → [20, 'Python']
# Manipulasi
data.append(99) # tambah di akhir
data.insert(1, 55) # sisipkan di posisi 1
data.remove(20) # hapus nilai pertama = 20

# data.sort() # urutkan (jika tipe sama)
```

## KESIMPULAN

Pada Kode ini, mendemonstrasikan dasar-dasar List Python: mulai dari pembuatan list dengan berbagai tipe data, mengakses elemen menggunakan indeks (termasuk slicing), hingga memanipulasi list dengan menambahkan elemen (append, insert) dan menghapusnya (remove). Ini adalah operasi fundamental dalam pengelolaan data menggunakan list di Python.
