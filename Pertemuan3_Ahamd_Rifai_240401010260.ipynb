```python
# ============================================
# PIPELINE DATA CLEANING — HOUSING DATASET
# ============================================
import pandas as pd, numpy as np
from scipy.stats.mstats import winsorize

# STEP 0 — Load & eksplorasi awal
df = pd.read_csv('housing_dirty.csv')
print('Shape awal:', df.shape)
print(df.isnull().sum())

# STEP 1 — Hapus Duplikat
df.drop_duplicates(inplace=True)
print('Setelah hapus duplikat:', df.shape)

# STEP 2 — Normalisasi String
df['kota'] = df['kota'].str.strip().str.title()
df['kondisi'] = df['kondisi'].str.strip().str.lower()

# STEP 3 — Imputasi Missing Values
df['luas_m2'] = df['luas_m2'].fillna(df['luas_m2'].median())
df['harga_juta'] = df['harga_juta'].fillna(df['harga_juta'].median())
df['kamar'] = df['kamar'].fillna(df['kamar'].mode()[0])
df['tahun_bangun'] = df['tahun_bangun'].fillna(df['tahun_bangun'].mode()[0])
df['kondisi'] = df['kondisi'].fillna(df['kondisi'].mode()[0])
df['kota'] = df['kota'].fillna(df['kota'].mode()[0])
#print('Setelah imputasi missing values:', df.shape)

# STEP 4 — Tangani Outlier (IQR Fence)
for col in ['harga_juta', 'luas_m2', 'tahun_bangun']:
  Q1, Q3 = df[col].quantile([0.25, 0.75])
  IQR = Q3 - Q1
  df[col] = df[col].clip(Q1-1.5*IQR, Q3+1.5*IQR)


# STEP 5 — Validasi & Ekspor
assert df.isnull().sum().sum() == 0, 'Masih ada missing!'
assert df.duplicated().sum() == 0, 'Masih ada duplikat!'
print('Shape akhir:', df.shape)
df.to_csv('housing_clean.csv', index=False)
print('Dataset bersih tersimpan!')
```

### Kesimpulan Data Cleaning : Missing, Outlier & Ekstraksi

Proses _data cleaning_ ini telah berhasil mengatasi beberapa masalah kualitas data pada housing_dirty. Langkah-langkah utama meliputi:

1.  **Penanganan _Missing Values_**: Mengisi nilai-nilai yang hilang pada kolom numerik dengan median dan pada kolom kategorikal dengan modus, memastikan kelengkapan data.
2.  **Penanganan _Outlier_**: Mengidentifikasi dan membatasi nilai-nilai ekstrem (_outlier_) pada kolom harga, luas, dan tahun bangun menggunakan metode IQR, untuk mengurangi distorsi analisis.
3.  **Normalisasi Data String**: Melakukan pembersihan dan standardisasi format teks pada kolom 'kota' dan 'kondisi' untuk memastikan konsistensi dan keseragaman data string.

Dengan demikian, _dataset_ lebih bersih, konsisten, dan siap untuk tahap analisis selanjutnya.
